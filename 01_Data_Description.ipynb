```python
# ------------------------------------------------------------
# 1. Load Data
# ------------------------------------------------------------

import pandas as pd

print("About to read the dataset...\n")

# Small dataframe with only two columns to prevent memory issues
columns = ["vat", "fail_dummy"]
df = pd.read_csv(
    "dataset.csv",
    usecols=columns,
    encoding="latin-1"
)

# Header-only dataframe
df_columns = pd.read_csv(
    "dataset.csv",
    nrows=0,
    encoding="latin-1"
)

print("Dataset successfully loaded.\n")
```

```python
# ------------------------------------------------------------
# 2. Dataset Overview
# ------------------------------------------------------------

n_observations = len(df)
n_variables = len(df_columns.columns)

print("DATASET OVERVIEW")
print("-" * 40)
print(f"Firm-year observations : {n_observations:,}")
print(f"Variables             : {n_variables:,}")
```

```python
# ------------------------------------------------------------
# 3. Firm-Level Class Distribution
# ------------------------------------------------------------

# Determine bankruptcy status at firm level
firms = df.groupby("vat")["fail_dummy"].max()

# Exclude firms with missing bankruptcy status
firms_known = firms.dropna()

n_firms = len(firms_known)
n_bankrupt = int(firms_known.sum())
n_non_bankrupt = n_firms - n_bankrupt
bankruptcy_rate = firms_known.mean() * 100

print(f"Firms with known status: {n_firms:,}")
print(f"Non-bankrupt firms: {n_non_bankrupt:,}")
print(f"Bankrupt firms: {n_bankrupt:,}")
print(f"Bankruptcy rate: {bankruptcy_rate:.2f}%")

print(f"\nTotal firms: {len(firms):,}")
print(f"Missing firm labels: {firms.isna().sum():,}")
print(f"Known firm labels: {firms.notna().sum():,}")
```

```python
# ------------------------------------------------------------
# 4. Firm-Year Class Distribution
# ------------------------------------------------------------

# Distribution of fail_dummy at the firm-year level
firm_year_distribution = df["fail_dummy"].value_counts(dropna=False)

# Number of labelled observations
n_labelled_obs = df["fail_dummy"].notna().sum()

# Number of bankrupt observations
n_bankrupt_obs = (df["fail_dummy"] == 1).sum()

# Bankruptcy rate among labelled observations
obs_bankruptcy_rate = (
    n_bankrupt_obs / n_labelled_obs
) * 100

print("\nFIRM-YEAR CLASS DISTRIBUTION")
print("-" * 40)

print(f"Total firm-year observations: {len(df):,}")
print(f"Labelled observations: {n_labelled_obs:,}")
print(f"Missing labels: {df['fail_dummy'].isna().sum():,}")

print("\nDistribution:")
print(firm_year_distribution)

print(f"\nBankrupt observations: {n_bankrupt_obs:,}")
print(
    f"Firm-year bankruptcy rate "
    f"(excluding missing labels): "
    f"{obs_bankruptcy_rate:.2f}%"
)
```

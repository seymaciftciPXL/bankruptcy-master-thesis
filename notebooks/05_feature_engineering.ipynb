```python
"""
05_feature_engineering.py

Build the financial ratio features used by the models:
- profitability, liquidity, solvency, structure, log_age
- save model_features.csv
"""
```

```python
from utils.load_data_selected import load_data_selected

df = load_data_selected()
```

```python
import numpy as np

from config import DATA_FEATURES, LOGS
from utils.print_section import print_section

print(df.shape)

# ------------------------------------------------------------------
# Financial ratios
# ------------------------------------------------------------------
print_section("Create financial ratios")

df["profitability"] = (df["net_income"] / df["ta"]).replace([np.inf, -np.inf], np.nan)
df["liquidity"] = (df["current_assets"] / df["st_debt"]).replace([np.inf, -np.inf], np.nan)
df["solvency"] = (df["rub10_15"] / df["ta"]).replace([np.inf, -np.inf], np.nan)
df["structure"] = (df["st_debt"] / df["total_debt"]).replace([np.inf, -np.inf], np.nan)
df["log_age"] = np.log1p(df["age_in_years"])

RATIO_COLUMNS = ["profitability", "liquidity", "solvency", "structure", "log_age"]

print(df[RATIO_COLUMNS].describe())

print_section("Missing values")
print(df[RATIO_COLUMNS].isna().sum())

# ------------------------------------------------------------------
# Save the dataset
# ------------------------------------------------------------------
print_section("Save feature dataset")

df.to_csv(DATA_FEATURES, index=False)
print(f"Saved to: {DATA_FEATURES}")

# ------------------------------------------------------------------
# Log
# ------------------------------------------------------------------
print_section("Write log")

LOG_PATH = LOGS / "feature_engineering_log.txt"

with open(LOG_PATH, "w") as f:
    f.write("FEATURE ENGINEERING LOG\n")
    f.write("=" * 60 + "\n\n")
    f.write(f"Rows: {len(df):,}\n")
    f.write(f"Columns: {len(df.columns):,}\n\n")
    f.write("Created features\n")
    f.write("-" * 60 + "\n")
    for col in RATIO_COLUMNS:
        f.write(f"{col}\n")

print(f"Log written to: {LOG_PATH}")
```
